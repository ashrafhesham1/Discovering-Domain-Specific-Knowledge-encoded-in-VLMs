```python
from PIL import Image
import io

shuffled_df = merged_df.sample(frac=1).reset_index(drop=True)
train_df = shuffled_df[:4000]
test_df = shuffled_df[4000:6000]
extra_data = shuffled_df[6000:]

def load_and_get_bytes(path):
    image = Image.open(f'/Users/shreyansh/Coursework/MMInterpretabilityResearch/Discovering-Domain-Specific-Knowledge-encoded-in-VLMs/data/VQA/val2014/COCO_val2014_{path:012d}.jpg')
    img_byte_arr = io.BytesIO()
    image.save(img_byte_arr, format='PNG')
    img_bytes = img_byte_arr.getvalue()
    return img_bytes

def get_image_df(df):
    path_arr = list(df["image_id"])
    image_bytes_arr = []
    for path in path_arr:
        image_bytes_arr.append(load_and_get_bytes(path))
    df['image'] = image_bytes_arr
    return df

train_df = get_image_df(train_df)
test_df = get_image_df(test_df)



train_df.to_parquet('train.parquet')
test_df.to_parquet('test.parquet')
```
